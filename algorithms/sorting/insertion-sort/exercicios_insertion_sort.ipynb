# Lista de Exercícios: Algoritmo Insertion Sort (Ordenação por Inserção)

### Breve Resumo do Insertion Sort
O Insertion Sort é um algoritmo simples e intuitivo de ordenação que constrói a lista ordenada final um elemento de cada vez. Ele funciona de maneira similar a como muitas pessoas ordenam cartas em um jogo de baralho:
1. Começamos a partir do segundo elemento (índice 1).
2. Comparamos o elemento atual com os elementos à sua esquerda.
3. Deslocamos todos os elementos maiores para a direita para abrir espaço.
4. Inserimos o elemento atual na sua posição correta.

**Complexidade:**
- **Pior Caso:** $O(n^2)$ (lista em ordem inversa)
- **Melhor Caso:** $O(n)$ (lista já ordenada)
- **Caso Médio:** $O(n^2)$
- **Espaço Auxiliar:** $O(1)$ (In-place/no próprio local)
- **Estabilidade:** Estável (preserva a ordem relativa de elementos iguais)

## Exercício 1: Implementação Básica do Insertion Sort

**Objetivo:** Implementar a função clássica do Insertion Sort que ordena uma lista de números em ordem crescente (do menor para o maior).

**Requisitos:**
1. A função `insertion_sort(arr)` deve modificar a lista original ou retornar uma nova lista ordenada.
2. Teste a sua implementação com a lista fornecida.

````python
def insertion_sort(arr):
    for i in range(len(arr)):
        aux = arr[i]
        j = i-1
        while (j > = 0 and arr[j] > aux):
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = aux

lista_teste = [12, 11, 13, 5, 6, 7]
print("Lista original:", lista_teste)

insertion_sort(lista_teste)
print("Lista ordenada:", lista_teste)
````

## Exercício 2: Ordenação Decrescente

**Objetivo:** Modifique o algoritmo de Insertion Sort para ordenar uma lista em ordem decrescente (do maior para o menor).

**Dica:** Observe a condição de comparação no laço `while`.

````python
def insertion_sort_decrescente(arr):
    for i in range(len(arr)):
        aux = arr[i]
        j = i-1
        while (j >= 0 and arr[j] < aux):
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = aux

# Teste
dados = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
print("Original:", dados)
insertion_sort_decrescente(dados)
print("Decrescente:", dados)
````

## Exercício 3: Rastreamento Passo a Passo (Visualização do Estado)

**Objetivo:** Modificar a função para que ela imprima o estado da lista a cada iteração do loop principal (cada vez que um elemento é inserido na sua posição correta). Isso ajuda a entender o funcionamento interno do algoritmo.

````python
def insertion_sort_passo_a_passo(arr):
    for i in range(len(arr)):
        aux = arr[i]
        j = i-1
        while j >= 0 and arr[j] > aux:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = aux
        print('Lista: i = {} j ={}: {}'.format(i,j,arr))
        

# Teste com uma lista de exemplo
lista_passo = [29, 10, 14, 37, 13]
insertion_sort_passo_a_passo(lista_passo)
````
