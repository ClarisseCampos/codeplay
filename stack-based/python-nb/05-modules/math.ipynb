## Exercício 1: Arredondamentos e Valor Absoluto

**Desafio:**  
Dada a nota final de um aluno `nota = 7.34` e a temperatura `temp = -12.8`:
1. Calcule o menor inteiro maior ou igual à nota
2. Calcule o maior inteiro menor ou igual à nota
3. Remova a parte decimal da nota para obter apenas a parte inteira
4. Calcule o valor absoluto da temperatura em ponto flutuante

```python
nota, temp = 7.34, -12.8

nota_cima = m.ceil(nota)
nota_baixo = m.floor(nota)
nota_truncada = m.trunc(nota)
temp_abs = m.fabs(temp)

# Exibindo os resultados
print(f"Ceil: {nota_cima}")
print(f"Floor: {nota_baixo}")
print(f"Trunc: {nota_truncada}")
print(f"Fabs: {temp_abs}")
```

## Exercício 3: Constantes e Trigonometria

**Desafio:**  
Em matemática, os ângulos para funções trigonométricas são dados em **radianos**.
1. Converta um ângulo de $45^\circ$ para radianos utilizando `math.radians`.
2. Calcule o **seno**, **cosseno** e a **tangente** desse ângulo em radianos usando `math.sin`, `math.cos` e `math.tan`.
3. Mostre o valor constante de $\pi$ (`math.pi`) e do número de Euler $e$ (`math.e`).

```python
angulo_graus = 45

angulo_rad = m.radians(angulo_graus)

seno = m.sin(angulo_rad)
cosseno = m.cos(angulo_rad)
tangente = m.tan(angulo_rad)

print(f"Ângulo em radianos: {angulo_rad:.4f}")
print(f"Seno: {seno:.4f} | Cosseno: {cosseno:.4f} | Tangente: {tangente:.4f}")
print(f"Constante Pi: {m.pi}")
print(f"Constante e: {m.e}")
```

## Exercício 4: Logaritmos e Teorema de Pitágoras

**Desafio:**  
1. Calcule o logaritmo natural ($\ln$) de $100$ (`math.log`).
2. Calcule o logaritmo de $100$ na base $10$ (`math.log10`).
3. Calcule a hipotenusa de um triângulo retângulo cujos catetos medem $a = 6$ e $b = 8$ usando a função `math.hypot`.

```python
log_nat = m.log(100)

log_10 = m.log10(100)

cateto_a = 6
cateto_b = 8
hipotenusa = m.sqrt(m.pow(cateto_a,2)+m.pow(cateto_b,2))

print(f"ln(100): {log_nat:.4f}")
print(f"log10(100): {log_10}")
print(f"Hipotenusa do triângulo (6, 8): {hipotenusa}")
```

## Exercício 5: Máximo Divisor Comum (MDC)

**Desafio:**  
Determine o Maior Divisor Comum (MDC) entre os números $48$ e $180$ utilizando a função.

```python
num1 = 48
num2 = 180

mdc = m.gcd(num1,num2)

print(f"O MDC entre {num1} e {num2} é: {mdc}")
```
