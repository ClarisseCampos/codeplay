# Strings


Strings in Python are arrays of bytes representing unicode characters. However, Python does not have a character data type, a single character is simply a string with a length of 1. Square brackets can be used to access elements of the string.

```python
character = 'Y'
print(type(character))
```

## Basic operators

### Concatenation

```python
q1 = "Hocus"
q2 = "Pocus"

print(q1,q2) # spaces
print(q1+q2) # no spaces

q3 = q1 +' '+ q2

print(q3)
```

```python
q = "Nyarlatothep"

print('arlato' in q)
```

```python
q = 'ha'
print(q*3)
```

### Indexing


```python
text = "Python"
print(text[0])  # Output: 'P'
print(text[3])  # Output: 'h'
```

### Case


```python
str = "good morning, dear Clarisse"

print(str.capitalize())
```

```python
str = input("Type something: ")

if(str.isupper()):
  print(f'The entire quote is upper case, so i will convet to you...{str.lower()}')

elif(str.islower()):
  print(f'The entire quote is lower case so i will convet to you...{str.upper()}')

else:
  print('Mixed-case quote')
```

```python
str = "SoMeThInG iN ThE wAy"
print(str.swapcase())
```

### Slicing

```python
qt = "Monty Python"

qt1 = qt[:5]
qt2 = qt[6:12]

print(qt1, qt2)
```
