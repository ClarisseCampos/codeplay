# Lista 1


1. Pegue os elementos da lista de trás para frente, pulando um.

```python
lista = [1,2,3,4,5,6,7,8]
print(lista[::-2])
```

2. Dada uma palavra, crie outra contendo apenas as letras em posições múltiplas de 3.

```python
s = "programacao"
s2 = s[::3]
print(s2)
```

3. Faça uma função que devolva apenas a primeira metade da string usando slicing.
Faça uma função que devolva apenas a segunda metade.

```python
s = "abcdef"
sz = len(s) // 2
s2 = s[:sz]
print(s2)
s3 = s[sz*(-1):]
print(s3)
```

4. Inverta somente os 5 primeiros caracteres.

```python
s = "computador"
print(s[-6::-1]+s[5:])
```
