# Dictionaries


```python
def report(player):
  print("\n=== # PLAYER'S REPORT # ===")
  print(f'* Name: {player['name']}')
  print(f'* Total games: {player['games']}')

  for i in range(player['games']):
    print(f'\t-> Wins in game x{i+1}: {player['wins'][i]}')

  print(f'* Total wins: {player['total']}')

player = {'name': input('Player name: '),
          'games': int(input('Q. games: ')),
          'wins': [],
          'total': 0
          }

for i in range(1,player['games']+1):
  player['wins'].append(int(input(f'Wins in game {i}: ')))
  player['total'] += player['wins'][-1]

report(player)
```
