```python
from manim import *
```

```python
%%manim -qm -v WARNING SquareAndCircle

class SquareAndCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.5)

        square = Square()
        square.set_fill(BLUE, opacity=0.5)

        triangle = Triangle()
        triangle.set_fill(GREEN, opacity=0.5)

        star = Star()
        star.set_fill(YELLOW, opacity=0.5)

        square.next_to(circle, RIGHT, buff=0.5)  # set the position
        triangle.next_to(circle, LEFT, buff=0.5)
        star.next_to(circle, DOWN, buff=0.5)

        self.play(Create(circle), Create(square), Create(triangle), Create(star))
```

```python
%%manim -qm -v WARNING MyScene

class MyScene(Scene):
    def construct(self):
        # Create the starting shape and add it to the scene
        shape = RegularPolygon(n=3)
        shape.set_fill(PINK, opacity=0.5)
        self.add(shape)

        for i in range(4, 9):
            next_shape = RegularPolygon(n=i)
            next_shape.set_fill(PINK, opacity=0.5)
            self.play(Transform(shape, next_shape))
            self.wait(0.5)
```
